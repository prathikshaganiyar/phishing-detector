```python
df['url_length'] = df['URL'].apply(len)
```

```python
df.head()
```

```python
df['count_dots'] = df['URL'].apply(lambda x: x.count('.'))
df['count_hyphens'] = df['URL'].apply(lambda x: x.count('-'))
```

```python
df['has_at_symbol'] = df['URL'].apply(lambda x: 1 if '@' in x else 0)
```

```python
import re

def has_ip(url):
    pattern = r'(\d{1,3}\.){3}\d{1,3}'
    if re.search(pattern, url):
        return 1
    else:
        return 0

df['has_ip'] = df['URL'].apply(has_ip)
```

```python
df.head()
```

```python
df['has_https'] = df['URL'].apply(lambda x: 1 if x.startswith('https') else 0)
```

```python
suspicious_words = ['login', 'verify', 'secure', 'account', 'update', 'confirm', 'banking', 'signin']

def count_suspicious_words(url):
    url_lower = url.lower()
    count = 0
    for word in suspicious_words:
        if word in url_lower:
            count += 1
    return count

df['suspicious_word_count'] = df['URL'].apply(count_suspicious_words)
```

```python
shortening_services = ['bit.ly', 'tinyurl.com', 'goo.gl', 't.co', 'ow.ly', 'is.gd', 'buff.ly']

def is_shortened(url):
    for service in shortening_services:
        if service in url:
            return 1
    return 0

df['is_shortened'] = df['URL'].apply(is_shortened)
```

```python
from urllib.parse import urlparse

def count_subdomains(url):
    try:
        domain = urlparse(url).netloc
        return domain.count('.')
    except:
        return 0

df['subdomain_count'] = df['URL'].apply(count_subdomains)
```

```python
def explain_prediction(url, model, feature_columns):
    # Extract features (same logic as your backend)
    features = {}
    features['url_length'] = len(url)
    features['count_dots'] = url.count('.')
    features['count_hyphens'] = url.count('-')
    features['has_at_symbol'] = 1 if '@' in url else 0

    ip_pattern = r'(\d{1,3}\.){3}\d{1,3}'
    features['has_ip'] = 1 if re.search(ip_pattern, url) else 0

    features['has_https'] = 1 if url.startswith('https') else 0

    suspicious_words = ['login', 'verify', 'secure', 'account', 'update', 'confirm', 'banking', 'signin']
    url_lower = url.lower()
    features['suspicious_word_count'] = sum(1 for word in suspicious_words if word in url_lower)

    shortening_services = ['bit.ly', 'tinyurl.com', 'goo.gl', 't.co', 'ow.ly', 'is.gd', 'buff.ly']
    features['is_shortened'] = 1 if any(service in url for service in shortening_services) else 0

    domain = urlparse(url).netloc
    features['subdomain_count'] = domain.count('.')

    feature_values = [features[col] for col in feature_columns]

    # Get prediction
    prediction = model.predict([feature_values])[0]
    probability = model.predict_proba([feature_values])[0]

    # Get feature importances (from Day 9)
    importances = model.feature_importances_

    print(f"URL: {url}")
    print(f"Prediction: {'Phishing' if prediction == 1 else 'Legitimate'}")
    print(f"Confidence: {max(probability)*100:.1f}%\n")
    print("Feature breakdown (value × importance):")
    for i, col in enumerate(feature_columns):
        contribution = feature_values[i] * importances[i]
        print(f"  {col}: value={feature_values[i]}, importance={importances[i]:.3f}, contribution={contribution:.3f}")

import re
from urllib.parse import urlparse

explain_prediction("https://www.amazon.com", rf_model, feature_columns)
```
